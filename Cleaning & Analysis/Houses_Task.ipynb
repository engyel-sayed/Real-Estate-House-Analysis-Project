# Q3
* ## Using aggregate function to calculate (Max, Min, Mean) of sales column 

```python
max_price = df["Price (£)"].max()
min_price = df["Price (£)"].min()
mean_price = df["Price (£)"].mean()
max_price, min_price, mean_price
```

# Q4
* ## Generate a column that specifies the price per square meter (.2F).
* ## Generte a column that specifies the total rooms in house

```python
df['Price_per_sqm'] = df['Price (£)'] / df['Square Meters']
df.head()
```

```python
df['Total_Rooms'] = df['Bedrooms'] + df['Bathrooms']
df.head()
```

# Q5
* ## What is the oldest house and price of it ?
* ## What is the newest house and price of it ?
### note(If there is more than one, calculate their mean.)

```python
max_age = df['Building Age'].max()
oldest_houses = df[df['Building Age'] == max_age]
oldest_houses
```

```python
mean_price_oldest = oldest_houses['Price (£)'].mean()
mean_price_oldest
```

```python
min_age = df['Building Age'].min()
newest_houses = df[df['Building Age'] == min_age]
newest_houses
```

```python
mean_price_newest = newest_houses['Price (£)'].mean()
mean_price_newest
```

# Q6
* ## Transform : (Garden , Garage) to numeric by mapping

#### note:(Yes = 1 / No = 0)

```python
mapping = {'Yes': 1, 'No': 0}
df['Garden'] = df['Garden'].map(mapping)
df['Garage'] = df['Garage'].map(mapping)
df.head()
```

# Q7
* ## Generate pivot tables for indexing by the following categories: (Neighborhood, Property Type, Heating Type, Balcony, View, Materials, and Building Status). The values will represent the mean price for each category.

* ## save each of them in a separate sheet within an Excel file 


```python
categories = ['Neighborhood', 'Property Type', 'Heating Type', 'Balcony', 'View', 'Materials', 'Building Status']
with pd.ExcelWriter('Houses_Analysis.xlsx') as writer:
    for cat in categories:
        pivot = df.pivot_table(index=cat, values='Price (£)', aggfunc='mean')
        pivot.to_excel(writer, sheet_name=cat)
```
